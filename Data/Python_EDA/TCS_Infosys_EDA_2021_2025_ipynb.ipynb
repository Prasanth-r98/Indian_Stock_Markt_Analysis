```python
def find_outliers(group):
    Q1 = group['Daily_Return'].quantile(0.25)
    Q3 = group['Daily_Return'].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return group[
        (group['Daily_Return'] < lower) |
        (group['Daily_Return'] > upper)
    ]

outliers = df.groupby(
    'Company',
    group_keys=False
).apply(find_outliers)

outliers[['Date', 'Company', 'Price', 'Daily_Return']].head(20)
```

```python
from scipy.stats import ttest_ind
```

```python
tcs_returns = df[
    df['Company'] == 'TCS'
]['Daily_Return'].dropna()

infosys_returns = df[
    df['Company'] == 'Infosys'
]['Daily_Return'].dropna()
```

```python
t_stat, p_value = ttest_ind(
    tcs_returns,
    infosys_returns,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("There is a statistically significant difference between TCS and Infosys daily returns.")
else:
    print("Fail to Reject H0")
    print("There is no statistically significant difference between TCS and Infosys daily returns.")
```
