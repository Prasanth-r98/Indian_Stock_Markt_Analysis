```python
import pandas as pd
df = pd.read_excel('TCS_Infosys_2021_2025_Combined.xlsx.xlsx')
print("Excel loaded successfully!")
```

```python
df.head()
```

```python
import sqlite3

conn = sqlite3.connect('stock_analysis.db')

print("SQL connection created successfully!")
```

```python
df.to_sql(
    'stock_data',
    conn,
    if_exists='replace',
    index=False
)

print("stock_data table created successfully!")
```

```python
query = """
SELECT *
FROM stock_data
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(*) AS Total_Records
FROM stock_data;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    COUNT(*) AS Total_Records
FROM stock_data
GROUP BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    ROUND(AVG(Price), 2) AS Average_Price
FROM stock_data
GROUP BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    MAX(Price) AS Highest_Price,
    MIN(Price) AS Lowest_Price
FROM stock_data
GROUP BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    ROUND(MAX(Price), 2) AS Highest_Price,
    ROUND(MIN(Price), 2) AS Lowest_Price
FROM stock_data
GROUP BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
pd.read_sql_query("PRAGMA table_info(stock_data);", conn)
```

```python
query = """
SELECT
    Company,
    ROUND(AVG(`Change %`), 4) AS Average_Change_Percent
FROM stock_data
GROUP BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    strftime('%Y', Date) AS Year,
    ROUND(AVG(Price), 2) AS Average_Price
FROM stock_data
GROUP BY Company, Year
ORDER BY Company, Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    strftime('%Y', Date) AS Year,
    ROUND(MAX(Price), 2) AS Maximum_Price
FROM stock_data
GROUP BY Company, Year
ORDER BY Company, Year;
"""

pd.read_sql_query(query, conn)
```

```python
pd.read_sql_query("PRAGMA table_info(stock_data);", conn)
```

```python
query = """
SELECT
    Date,
    Company,
    Price,
    `Change %` AS Change_Percent
FROM stock_data
ORDER BY `Change %` DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Date,
    Company,
    Price,
    `Change %` AS Change_Percent
FROM stock_data
ORDER BY `Change %` ASC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    Date,
    Price,
    `Change %` AS Change_Percent
FROM stock_data
WHERE `Change %` = (
    SELECT MAX(s2.`Change %`)
    FROM stock_data s2
    WHERE s2.Company = stock_data.Company
)
ORDER BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Company,
    Date,
    Price,
    `Change %` AS Change_Percent
FROM stock_data
WHERE `Change %` = (
    SELECT MIN(s2.`Change %`)
    FROM stock_data s2
    WHERE s2.Company = stock_data.Company
)
ORDER BY Company;
"""

pd.read_sql_query(query, conn)
```

```python
pd.read_sql_query("PRAGMA table_info(stock_data);", conn)
```

```python
query = """
SELECT
    Company,
    COUNT(*) AS Total_Records,
    ROUND(AVG(Price), 2) AS Average_Price,
    ROUND(MAX(Price), 2) AS Highest_Price,
    ROUND(MIN(Price), 2) AS Lowest_Price,
    SUM(CAST(REPLACE(`Vol.`, ',', '') AS REAL)) AS Total_Volume,
    ROUND(AVG(`Change %`), 4) AS Average_Change_Percent
FROM stock_data
GROUP BY Company
ORDER BY Company;
"""

final_sql_summary = pd.read_sql_query(query, conn)

final_sql_summary
```

```python
final_sql_summary.to_excel(
    'SQL_Final_Summary.xlsx',
    index=False
)

print("SQL Final Summary saved successfully!")
```
